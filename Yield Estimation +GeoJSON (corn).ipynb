```python
import os
import glob
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ultralytics import YOLO
from pathlib import Path
from tqdm import tqdm

# -------------------------------------------------
# CONFIGURATION – change these 3 paths only
# -------------------------------------------------
MODEL_PATH = r"corn_best(12).pt"
INPUT_DIR = r""
OUTPUT_DIR = ""
# -------------------------------------------------
# Constants for plant stand estimation
PLANTS_PER_LINEAR_FT = 2  # plant to plant 6 inches, row to row 24 inches (confirm that from Hassan )
ROW_SPACING_INCHES = 24 
FIELD_ACRES = 130
IMAGE_COUNT = 167
CAMERA_HEIGHT_M = 12  # drone flight height
IMAGE_RESOLUTION = (8192, 6144)  # width, height

#Yield Constants:
Lbs_per_Plant = 1.5   # average weight of Corn plant for silage 65% moisture (Pick seed 223)
Lb_per_ton=2240  # pounds per ton
harvest_loss_percentage=0.00 # 5% loss during harvest

# Calculate plant density
row_spacing_ft = ROW_SPACING_INCHES / 12
plants_per_sqft = PLANTS_PER_LINEAR_FT / row_spacing_ft
plants_per_acre = plants_per_sqft * 43560
total_plants = plants_per_acre * FIELD_ACRES

# Calculate image coverage
# 24mm equivalent lens (DJI Air 3S)
horizontal_fov = 69.5  # degrees
vertical_fov = 54.4  # degrees

# Calculate ground coverage per image
ground_width = 2 * CAMERA_HEIGHT_M * np.tan(np.radians(horizontal_fov / 2))
ground_height = 2 * CAMERA_HEIGHT_M * np.tan(np.radians(vertical_fov / 2))
area_per_image = ground_width * ground_height  # m²

# Convert to acres (1 acre = 4046.86 m²)
area_per_image_acres = area_per_image / 4046.86
total_coverage_acres = area_per_image_acres * IMAGE_COUNT
coverage_percentage = (total_coverage_acres / FIELD_ACRES) * 100

os.makedirs(OUTPUT_DIR, exist_ok=True)
# Load YOLO model
model = YOLO(MODEL_PATH)
Confidence_threshold = 0.1
# Container for final counts
results = []
image_files = list(Path(INPUT_DIR).rglob("*"))
image_files = [str(p) for p in image_files if p.suffix.lower() in {".jpg", ".jpeg", ".png"}]

for img_path in tqdm(image_files, desc="Processing images"):
    img = cv2.imread(str(img_path))
    if img is None:
        continue

    preds = model(img, conf=Confidence_threshold,iou=0.5, max_det=5000)[0]

    # 🔁 apply confidence threshold
    boxes = [b for b in preds.boxes if b.conf.item() > Confidence_threshold]
    plant_count = len(boxes)

    annotated= img.copy()
    for box in boxes:
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        cv2.rectangle(annotated, (x1, y1), (x2, y2), (0, 255, 210), 2)
        cv2.putText(annotated, f"Corn_V4", (x1, y1 - 5),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (150, 255, 0), 2)
    out_img = os.path.join(OUTPUT_DIR, os.path.basename(img_path))
    cv2.imwrite(str(out_img), annotated)

    # ➕ store plants/m² as well
    plants_per_m2 = plant_count / area_per_image
    results.append({
        "image": Path(img_path).name,
        "plants": plant_count,
        "plants_per_m2": plants_per_m2
    })

# Optional CSV
df = pd.DataFrame(results)
csv_path = Path(OUTPUT_DIR) / "plant_counts.csv"
df.to_csv(csv_path, index=False)
total_detected = df['plants'].sum()
predicted_total = total_detected * (100 / coverage_percentage)
plants_predicted_per_acre = predicted_total / FIELD_ACRES

def plants_to_yield(plants_per_acre):
    total_weight_lbs = plants_per_acre * Lbs_per_Plant
    total_weight_lbs_after_loss = total_weight_lbs * (1 - harvest_loss_percentage)
    return total_weight_lbs_after_loss/Lb_per_ton

target_yield_per_acre_tons = plants_to_yield(plants_per_acre)
detected_yield_per_acre_tons = plants_to_yield(plants_predicted_per_acre)
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 12))
# ------------------------------------------------------------------
labels = ['Covered by Images', 'Remaining Field']
sizes = [coverage_percentage, 100 - coverage_percentage]
colors = ['#ff9999', '#66b3ff']
ax1.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
ax1.set_title('Field Coverage')
# Build concise summary text for inside the pie subplot
summary_txt = (
    f"Field: {FIELD_ACRES} acres\n"
    f"Images: {len(df)}\n"
    f"Coverage: {coverage_percentage:.2f} %\n"
    f"Est. Total: {total_plants:,.0f}\n"
    f"Pred. Total: {predicted_total:,.0f}"
)
ax1.annotate(summary_txt,
             xy=(0.5, 0.5),  # centre of pie
             xycoords='axes fraction',
             ha='center', va='center',
             fontsize=10,
             bbox=dict(boxstyle='round,pad=0.4',
                       facecolor='white',
                       edgecolor='gray',
                       alpha=0.9))
# ------------------------------------------------------------------
# 2) Total plant count comparison (TOP-RIGHT)
# ------------------------------------------------------------------
cats_total = ['Targeted', 'detected ']
vals_total = [total_plants, predicted_total]
ax2.bar(cats_total, vals_total, color=['#1f77b4', '#ff7f0e'])
ax2.set_title('Total Plant Count – 130 acres')
ax2.set_ylabel('Number of Plants')
for i, v in enumerate(vals_total):
    ax2.text(i, v + max(vals_total) * 0.01, f'{v:,.0f}', ha='center')
# ------------------------------------------------------------------
# 3) Plants-per-acre comparison (BOTTOM-LEFT)
# ------------------------------------------------------------------
cats_acre = ['Targeted', 'detected']
vals_acre = [plants_per_acre, plants_predicted_per_acre]
ax3.bar(cats_acre, vals_acre, color=["#6a6060", '#ff7f0e'])
ax3.set_title('Plants per Acre Comparison')
ax3.set_ylabel('Plants / Acre')
for i, v in enumerate(vals_acre):
    ax3.text(i, v + max(vals_acre) * 0.01, f'{v:,.0f}', ha='center')
# ------------------------------------------------------------------
# 4) Plants-per-image histogram (BOTTOM-RIGHT)
# ------------------------------------------------------------------
# ------ 4) NEW: Yield comparison (US ton per acre) ------
cats_yield = ['Targeted\nYield', 'Detected\nYield']
vals_yield = [target_yield_per_acre_tons, detected_yield_per_acre_tons]
bars = ax4.bar(cats_yield, vals_yield, color=['#2ca02c', "#27aad6"])
ax4.set_title(' Field Estimated Yield')
ax4.set_ylabel('ton / acre')
for i, v in enumerate(vals_yield):
    ax4.text(i, v + max(vals_yield) * 0.01, f'{v:.1f}', ha='center')
# ------------------------------------------------------------------
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, 'summary_analysis.png'), dpi=300)
plt.show()
# Optional console echo
print("\nSummary Analysis:")
print(f"Coverage: {coverage_percentage:.2f}% | "
      f"Observed in images: {df['plants'].sum():,} | "
      f"Predicted total: {predicted_total:,.0f}")
```

### GeoJSON


```python
"""
Only export plants.geojson – no CSV, no PDF, no images.
"""
import json, math, cv2, os
from pathlib import Path
import exifread
from tqdm import tqdm
from ultralytics import YOLO

# -------------------------------------------------
# USER PATHS

MODEL_PATH = Path(r"corn_best(12).pt")
INPUT_DIR = Path(r"FM_Corn_V4_Enchant_WF_20250605_12m_A3S")
OUTPUT_DIR = Path(r"Result_folder/FM_Corn_V4_Enchant_WF_20250605_12m_A3S(results-1)")
GEOJSON_PATH = OUTPUT_DIR / "Corn(1).geojson"

# ---------------- CONSTANTS --------------------
CAMERA_HEIGHT_M = 12          # flight height
horizontal_fov  = 69.5        # deg
CONF_THR        = 0.1

# ---------------- HELPERS ----------------------
def exif_gps(path):
    with open(path, 'rb') as f:
        t = exifread.process_file(f, details=False)
    for tag in ('GPS GPSLatitude', 'GPS GPSLongitude', 'GPS GPSLatitudeRef', 'GPS GPSLongitudeRef'):
        if tag not in t: return None, None
    lat = sum(float(x.num)/float(x.den)/60**i for i, x in enumerate(t['GPS GPSLatitude'].values))
    lon = sum(float(x.num)/float(x.den)/60**i for i, x in enumerate(t['GPS GPSLongitude'].values))
    if t['GPS GPSLatitudeRef'].values  != 'N': lat = -lat
    if t['GPS GPSLongitudeRef'].values != 'E': lon = -lon
    return lat, lon

def pixel_to_geo(px, py, img_w, img_h, lat, lon, fov_h_deg, alt_m):
    if lat is None: return None, None
    gsd = (2 * alt_m * math.tan(math.radians(fov_h_deg / 2))) / img_w
    dx  = (px - img_w / 2) * gsd
    dy  = (img_h / 2 - py) * gsd
    lat_off = dy / 111320
    lon_off = dx / (111320 * math.cos(math.radians(lat)))
    return lat + lat_off, lon + lon_off

# --------------- IMAGE LIST ------------------
images = [str(p) for p in INPUT_DIR.rglob('*') if p.suffix.lower() in {'.jpg','.jpeg','.png'}]

# --------------- MODEL -----------------------
model = YOLO(str(MODEL_PATH))

# --------------- GEOJSON ONLY ----------------
features = []
img_id = 0
imgs_to_process=images[::2]
for img_path in tqdm(images, desc="GeoJSON only"):
    img = cv2.imread(img_path)
    if img is None: continue

    preds = model(img, conf=CONF_THR, max_det=4000)[0]
    h, w  = img.shape[:2]
    lat_d, lon_d = exif_gps(img_path)
    img_id += 1
    img_name = Path(img_path).name

    for idx, box in enumerate(preds.boxes):
        if float(box.conf[0]) <= CONF_THR: continue
        x1,y1,x2,y2 = box.xyxy[0].tolist()
        cx = (x1 + x2) / 2
        cy = (y1 + y2) / 2
        lat_p, lon_p = pixel_to_geo(cx, cy, w, h, lat_d, lon_d, horizontal_fov, CAMERA_HEIGHT_M)
        features.append({
            "type": "Feature",
            "geometry": {"type": "Point",
                         "coordinates": [lon_p if lon_p is not None else 0,
                                         lat_p if lat_p is not None else 0]},
            "properties": {"image": img_name,
                           "plant_id": f"{img_id}_{idx}",
                           "confidence": round(float(box.conf[0]), 3),
                           "gps_valid": lat_p is not None}
        })

# --------------- SAVE GEOJSON ---------------
geojson = {"type": "FeatureCollection", "features": features}
with open(GEOJSON_PATH, 'w', encoding='utf-8') as f:
    json.dump(geojson, f, indent=2)
print(f"GeoJSON saved → {GEOJSON_PATH}  ({len(features)} plants)")
```
